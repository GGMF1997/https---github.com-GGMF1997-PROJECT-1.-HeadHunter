3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def get_experience_month(arg):
    """Преобразует текстовое описание опыта работы в число месяцев.

    Примеры входных строк: «Опыт работы 16 лет 10 месяцев ...».
    Возвращает NaN, если значение отсутствует или «Не указано».
    """
    if arg is np.nan or arg == 'Не указано':
        return np.nan

    year_words = ['год', 'года', 'лет']
    month_words = ['месяц', 'месяца', 'месяцев']

    # Достаточно первых 7 токенов, где обычно указана длительность
    arg_split = arg.split(' ')[:7]

    years = 0
    months = 0

    for i, value in enumerate(arg_split):
        if value in year_words:
            years = int(arg_split[i - 1])
        if value in month_words:
            months = int(arg_split[i - 1])

    return years * 12 + months


# Создаём числовой признак опыта в месяцах и удаляем исходный текст
hh_df['Опыт работы (месяц)'] = hh_df['Опыт работы'].apply(get_experience_month)
hh_df = hh_df.drop('Опыт работы', axis=1)
```

```python
# Медианный опыт работы в месяцах
hh_df['Опыт работы (месяц)'].median()
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
def town_type(arg):
    """Классифицирует город соискателя.

    Возвращает одну из категорий:
    «Москва», «Санкт-Петербург», «город-миллионник» или «другие».
    """
    million_cities = [
        'Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань',
        'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа',
        'Красноярск', 'Пермь', 'Воронеж', 'Волгоград',
    ]
    city = arg.split(' ')[0]

    if city == 'Москва':
        return 'Москва'
    if city == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    if city in million_cities:
        return 'город-миллионник'
    return 'другие'


hh_df['Город'] = hh_df['Город, переезд, командировки'].apply(town_type)
```

```python
# Доля соискателей по категориям городов (в процентах)
hh_df['Город'].value_counts(normalize=True) * 100
```

```python
def relocation_update(arg):
    """True, если соискатель готов к переезду, иначе False."""
    if ('не готов к переезду' in arg) or ('не готова к переезду' in arg):
        return False
    return True


hh_df['Готовность к переезду'] = (
    hh_df['Город, переезд, командировки'].apply(relocation_update)
)
```

```python
def business_trips(arg):
    """True, если соискатель готов к командировкам, иначе False."""
    if ('не готов к командировкам' in arg) or ('не готова к командировкам' in arg):
        return False
    return True


hh_df['Готовность к командировкам'] = (
    hh_df['Город, переезд, командировки'].apply(business_trips)
)
```

```python
# Доля соискателей, готовых и к переезду, и к командировкам
mask = (
    (hh_df['Готовность к командировкам'] == True)
    & (hh_df['Готовность к переезду'] == True)
)
mask.sum() / hh_df.shape[0]
```

```python
hh_df = hh_df.drop('Город, переезд, командировки', axis=1)
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
# Бинарные признаки типов занятости
employment_list = [
    'полная занятость',
    'частичная занятость',
    'проектная работа',
    'волонтерство',
    'стажировка',
]
for value in employment_list:
    hh_df[value] = hh_df['Занятость'].apply(lambda x: value in x)

# Бинарные признаки графиков работы
schedule_list = [
    'полный день',
    'сменный график',
    'гибкий график',
    'удалённая работа',
    'вахтовый метод',
]
for value in schedule_list:
    hh_df[value] = hh_df['График'].apply(lambda x: value in x)
```

```python
# Число резюме с одновременной готовностью к проектной работе и волонтёрству
((hh_df['проектная работа'] == True) & (hh_df['волонтерство'] == True)).sum()
```

```python
# Число резюме с вахтовым методом и гибким графиком одновременно
((hh_df['вахтовый метод'] == True) & (hh_df['гибкий график'] == True)).sum()
```

```python
# Курсы валют для перевода зарплат в рубли
currency_data = pd.read_csv('data/ExchangeRates.csv', sep=',')
```

```python
# Приводим даты курсов и обновления резюме к типу date
currency_data['date'] = pd.to_datetime(currency_data['date'], dayfirst=True)
currency_data['date'] = currency_data['date'].dt.date
```

```python
# Какие валюты присутствуют в справочнике курсов
currency_data['currency'].value_counts()
```

```python
# Дата обновления резюме (только дата, без времени)
hh_df['Обновление резюме'] = pd.to_datetime(
    hh_df['Обновление резюме'], dayfirst=True
)
hh_df['Обновление резюме'] = hh_df['Обновление резюме'].dt.date
```

```python
# Разбиваем столбец «ЗП» на сумму и название валюты
hh_df['Сумма желаемой заработной платы'] = hh_df['ЗП'].apply(
    lambda x: x.split(' ')[0]
)
hh_df['Наименование валюты'] = hh_df['ЗП'].apply(lambda x: x.split(' ')[1])
hh_df['Наименование валюты'] = hh_df['Наименование валюты'].apply(
    lambda x: x.replace('.', '')
)
```

```python
# Частота валют в желаемой зарплате (до приведения к ISO)
hh_df['Наименование валюты'].value_counts()
```

```python
def to_iso(currency):
    """Переводит локальное название валюты в код ISO 4217."""
    mapping = {
        'грн': 'UAH',
        'USD': 'USD',
        'EUR': 'EUR',
        'белруб': 'BYN',
        'KGS': 'KGS',
        'сум': 'UZS',
        'AZN': 'AZN',
        'KZT': 'KZT',
        'руб': 'RUB',
    }
    return mapping.get(currency, currency)


hh_df['Наименование валюты'] = hh_df['Наименование валюты'].apply(to_iso)
```

```python
# Частота валют после приведения к ISO-кодам
hh_df['Наименование валюты'].value_counts()
```

```python
# Объединяем резюме с курсами по дате обновления и валюте
hh_merged = hh_df.merge(
    currency_data,
    left_on=['Обновление резюме', 'Наименование валюты'],
    right_on=['date', 'currency'],
    how='left',
)
```

```python
# Для рублёвых зарплат курса нет — подставляем 1
hh_merged['close'] = hh_merged['close'].fillna(1)
hh_merged['proportion'] = hh_merged['proportion'].fillna(1)
```

```python
# Сумма желаемой ЗП — целое число
hh_merged['Сумма желаемой заработной платы'] = (
    hh_merged['Сумма желаемой заработной платы'].astype(int)
)
```

```python
# Пересчёт желаемой зарплаты в рубли с учётом курса и номинала
hh_merged['ЗП (руб)'] = (
    hh_merged['Сумма желаемой заработной платы']
    * hh_merged['close']
    / hh_merged['proportion']
)
```

```python
# Медианная желаемая зарплата в рублях
hh_merged['ЗП (руб)'].median()
```

```python
# Возвращаем результат в основной датафрейм
hh_df = hh_merged
```

```python

def ouliers_iqr(data, feature):
    """Пишем функцию для поиска выбросов с помощью метода Тьюки"""
    x = data[feature]
    quartile_1, quartile_3 = x.quantile(0.25), x.quantile(0.75),
    iqr = quartile_3 - quartile_1
    lower_bound = quartile_1 - (iqr * 1.5)
    upper_bound = quartile_3 + (iqr * 1.5)
    outliers = data[(x < lower_bound) | (x > upper_bound)]
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)] 
    return outliers, cleaned

outliers, cleaned = ouliers_iqr(hh_df, 'ЗП (руб)')
```

```python
# Печатаем кол-во строк с выбросами и очищенными данными
print(outliers.shape[0])
print(cleaned.shape[0])
```

6. В результате анализа мы обнаружили потенциальные выбросы в признаке **"Возраст"**. Это оказались резюме людей чересчур преклонного возраста для поиска работы. Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
def outliers_z_score(
    data,
    feature,
    log_scale=False,
    left=3,
    right=4,
):
    """Поиск выбросов методом z-оценки.

    data : DataFrame
        Исходные данные.
    feature : str
        Имя числового признака.
    log_scale : bool
        Если True, применяется логарифмирование.
    left, right : float
        Число стандартных отклонений слева и справа от среднего.

    Возвращает
    outliers, cleaned : DataFrame
        Выбросы и очищенные данные.
    """
    if log_scale:
        x = np.log(data[feature])
    else:
        x = data[feature]

    sigma = x.std()
    mu = x.mean()
    lower_bound = mu - left * sigma
    upper_bound = mu + right * sigma

    outliers = data[(x < lower_bound) | (x > upper_bound)]
    cleaned = data[(x > lower_bound) & (x < upper_bound)]
    return outliers, cleaned


# Выбросы по возрасту (логарифмическая шкала)
outliers, cleaned = outliers_z_score(hh_df, 'Возраст', log_scale=True)
```

```python
# Гистограмма возраста с границами z-оценки
age = cleaned['Возраст']
mu = age.mean()
sigma = age.std()

fig, ax = plt.subplots(figsize=(10, 4))
age_hist = sns.histplot(data=cleaned, x='Возраст')
age_hist.axvline(mu, color='r', lw=2)
age_hist.axvline(mu - 3 * sigma, color='b', lw=2)
age_hist.axvline(mu + 4 * sigma, color='g', lw=2)
age_hist.set_title('Гистограмма распределения возраста')
```
